```python
!pip install nltk

# Import required libraries
import nltk
from nltk.corpus import wordnet
from nltk.tokenize import word_tokenize

# Download necessary datasets
nltk.download('punkt')
nltk.download('wordnet')
```

```python
query = input("Enter your query: ")
print("Original Query:", query)
```

```python
tokens = word_tokenize(query)
print("Tokenized Words:", tokens)
```

```python
expanded_query = []
```

```python
for word in tokens:
    # Add original word
    expanded_query.append(word)

    # Get synonyms
    synonyms = set()

    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name())

    # Add synonyms to expanded query
    expanded_query.extend(synonyms)

    # Print step output
    print(f"\nWord: {word}")
    print("Synonyms Found:", list(synonyms))
```

```python
expanded_query = list(set(expanded_query))
print("\nExpanded Query Terms:", expanded_query)
```

```python
final_query = " ".join(expanded_query)
print("\nFinal Expanded Query:")
print(final_query)
```
